```python
# Your Phase 1 solution to the problem goes here.

from google.colab import files
uploaded = files.upload()
import pandas as pd
import numpy as np

#Q1.1

df = pd.read_csv('airbnb_hw.csv')

print(df.head())
print(df.shape)
print(df.info)
#Overview of the dataframe

df['Price'] = df['Price'].str.replace(',', '', regex=False)
df['Price'] = pd.to_numeric(df['Price'], errors='coerce')
df = df.sort_values('Price')
# remove commas from 4+ digit price values
# convert to numeric values fron strings
# sort the dataset by price values

print('Airbnb Price Cleaning Exercise')
print(f"Missing values: {df['Price'].isna().sum()}")
# check for missing values

print(df['Price'].describe())
# verify data cleaning




#Q1.2
df2 = pd.read_csv('mn_police_use_of_force.csv')

print(df2.head())
print(df2.shape)
print(df2.info())
# Overview of the dataframe

# subject_injury is already clean - just "Yes", "No", and NaN

print('Minnesota Police Use of Force - subject_injury Analysis')
print(f"Missing values: {df2['subject_injury'].isna().sum()}")
print(f"Percentage missing: {(df2['subject_injury'].isna().sum() / len(df2) * 100):.2f}%")
# check for missing values

print("\nValue counts:")
print(df2['subject_injury'].value_counts(dropna=False))
# see breakdown of Yes, No, and missing

print("\nCrosstab: subject injury vs force type")
crosstab = pd.crosstab(df2['subject_injury'], df2['force_type'], margins=True, margins_name='Total')
print(crosstab)
# finds patterns in missing data




# Q2

```
