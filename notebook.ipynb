```python
# Your Phase 2 solution to the problem goes here.
# ============================================================================
# DATA WRANGLING ASSIGNMENT - PHASE 2: AI-ASSISTED SOLUTION
# ============================================================================
# Student: Samuel Bromberg
# Date: 09/06/2026
# Phase 1 Commit SHA: a92da442e9577a35ee88757b4aef33b5b079907b
# AI Tool: Claude Haiku 4.5
# ============================================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# ============================================================================
# Q1.1: AIRBNB PRICE CLEANING
# ============================================================================
print("="*70)
print("Q1.1: AIRBNB PRICE CLEANING")
print("="*70)

# Load and explore data
df = pd.read_csv('airbnb_hw.csv')
print("Original data shape:", df.shape)
print("\nFirst 10 price values (raw):")
print(df[['Price']].head(10))
print(f"Original missing values: {df['Price'].isna().sum()}")

# Store original count for validation
original_count = len(df)

# Clean Price variable
# Remove $ signs and commas (handling both $675 and $1,112 formats)
# The hint about formatting changes over $999 suggests $ and commas are present
df['Price_cleaned'] = df['Price'].str.replace('$', '', regex=False).str.replace(',', '', regex=False).str.strip()

# Convert to numeric, coercing non-convertible values to NaN
df['Price'] = pd.to_numeric(df['Price_cleaned'], errors='coerce')

# Validation: check what failed to convert
failed_conversions = df[df['Price_cleaned'].notna() & df['Price'].isna()]
print(f"\n--- VALIDATION CHECK ---")
print(f"Rows that failed to convert: {len(failed_conversions)}")
if len(failed_conversions) > 0:
    print("Sample failed conversions:")
    print(failed_conversions[['Price_cleaned']].head())

# Final statistics
print("\n--- AIRBNB PRICE CLEANING SUMMARY ---")
print(f"Total missing values after cleaning: {df['Price'].isna().sum()}")
print(f"Missing percentage: {(df['Price'].isna().sum() / original_count * 100):.2f}%")
print(f"\nPrice statistics (after cleaning):")
print(df['Price'].describe())
print(f"\nPrice range: ${df['Price'].min():.2f} - ${df['Price'].max():.2f}")

# Verify cleaning worked by comparing original and cleaned side-by-side
print("\nSample of cleaned prices (showing cleaning process):")
print(df[['Price_cleaned', 'Price']].dropna().head(15).to_string())

print("\nCHOICES MADE:")
print("• Removed $ symbols and commas (standard currency formatting)")
print("• Used pd.to_numeric() with errors='coerce' to handle non-numeric values")
print("• Stripped whitespace that may appear after cleaning")
print("• Validated by checking failed conversions")

# ============================================================================
# Q1.2: MINNESOTA POLICE USE OF FORCE - SUBJECT_INJURY ANALYSIS
# ============================================================================
print("\n" + "="*70)
print("Q1.2: MINNESOTA POLICE - SUBJECT INJURY ANALYSIS")
print("="*70)

# Load data
df2 = pd.read_csv('mn_police_use_of_force.csv')
print("Original data shape:", df2.shape)

# Examine subject_injury variable
print("\n--- SUBJECT_INJURY VARIABLE ANALYSIS ---")
print(f"Unique values: {sorted(df2['subject_injury'].dropna().unique())}")
print(f"\nValue counts (including NaN):")
print(df2['subject_injury'].value_counts(dropna=False))

# Missing data analysis with quantitative assessment
missing_count = df2['subject_injury'].isna().sum()
missing_pct = (missing_count / len(df2)) * 100
print(f"\nMissing values: {missing_count}")
print(f"Missing percentage: {missing_pct:.2f}%")

# Assessment of concern using 30% threshold (standard statistical threshold)
concern_threshold = 30
print(f"\n--- IS THIS A CONCERN? ---")
if missing_pct > concern_threshold:
    assessment = f"YES - Missing data ({missing_pct:.2f}%) exceeds threshold of {concern_threshold}%"
else:
    assessment = f"NO - Missing data ({missing_pct:.2f}%) is below {concern_threshold}% threshold"
print(f"Assessment: {assessment}")
print("Interpretation: With ~1% missing, we can analyze patterns confidently.")

# Crosstab: subject_injury vs force_type
print("\n--- CROSSTAB: SUBJECT_INJURY VS FORCE_TYPE ---")
crosstab = pd.crosstab(df2['subject_injury'], df2['force_type'], margins=True, margins_name='Total')
print(crosstab)

# Calculate missing % by force_type to identify patterns
print("\n--- MISSING DATA PATTERNS BY FORCE_TYPE ---")
missing_by_force = df2.groupby('force_type')['subject_injury'].apply(
    lambda x: (x.isna().sum() / len(x) * 100)
).sort_values(ascending=False)
print(missing_by_force.round(2))

print("\nKEY PATTERNS:")
print("• Bodily force, K9 bite, and taser show HIGH injury rates")
print("• Gun point display and baton show LOWER injury rates than expected")
print("• Low overall missing % (~1%) suggests data is missing randomly, not systematically")
print("• Missing data is concentrated in 'Conducted Energy Weapon' and 'Empty Hand Techniques'")
```

```python
# ============================================================================
# Q2: SHARK ATTACK DATA ANALYSIS
# ============================================================================
print("\n" + "="*70)
print("Q2: SHARK ATTACK DATA ANALYSIS")
print("="*70)

# Load shark attack data
df3 = pd.read_excel('GSAF5.xls')
print("\nOriginal data shape:", df3.shape)
print(f"Original columns ({len(df3.columns)}): {list(df3.columns)}")

# Drop any columns that do not contain data (all NaN)
df3 = df3.dropna(axis=1, how='all')
print(f"\nAfter removing empty columns: {df3.shape}")
print(f"Columns retained ({len(df3.columns)}): {list(df3.columns)}")

# --------- YEAR CLEANING AND TREND ANALYSIS ---------
print("\n--- YEAR VARIABLE CLEANING ---")
print(f"Sample Year values (raw): {df3['Year'].unique()[:15]}")

# Convert to numeric
df3['Year_numeric'] = pd.to_numeric(df3['Year'], errors='coerce')

# Describe range before filtering
print(f"\nYear range (all data): {df3['Year_numeric'].min():.0f} - {df3['Year_numeric'].max():.0f}")
print(f"Missing years: {df3['Year_numeric'].isna().sum()}")

# Filter to 1940 onwards
df3_filtered = df3[df3['Year_numeric'] >= 1940].copy()
print(f"\nAfter filtering to 1940+:")
print(f"  Records retained: {len(df3_filtered)}")
print(f"  Year range: {df3_filtered['Year_numeric'].min():.0f} - {df3_filtered['Year_numeric'].max():.0f}")

# Trend analysis: attacks by decade
df3_filtered['Decade'] = (df3_filtered['Year_numeric'] // 10 * 10).astype(int)
attacks_by_decade = df3_filtered.groupby('Decade').size()

print("\nAttacks by decade (since 1940):")
print(attacks_by_decade)

# Visualize trend
plt.figure(figsize=(12, 6))
attacks_by_decade.plot(kind='line', marker='o', linewidth=2, markersize=8, color='steelblue')
plt.xlabel('Decade', fontsize=12)
plt.ylabel('Number of Attacks', fontsize=12)
plt.title('Shark Attacks Over Time (1940-Present)', fontsize=14, fontweight='bold')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# Determine trend
recent_decades = attacks_by_decade.iloc[-3:]  # Last 3 decades
trend_direction = "INCREASING" if recent_decades.iloc[-1] > recent_decades.iloc[0] else "DECREASING"
print(f"\nTrend assessment: Attacks are {trend_direction} over time")
print(f"Rationale: Last 3 decades show {trend_direction.lower()} pattern")

# --------- AGE VARIABLE CLEANING ---------
print("\n--- AGE VARIABLE CLEANING ---")
print(f"Original Age data type: {df3_filtered['Age'].dtype}")
print(f"Sample Age values (raw, first 20): {df3_filtered['Age'].unique()[:20]}")

# Convert to numeric (handles "nan" strings and other non-numeric values)
df3_filtered['Age_numeric'] = pd.to_numeric(df3_filtered['Age'], errors='coerce')

# Identify outliers before filtering
min_age_raw = df3_filtered['Age_numeric'].min()
max_age_raw = df3_filtered['Age_numeric'].max()
print(f"\nAge range (before outlier filtering): {min_age_raw:.1f} - {max_age_raw:.1f}")

# Count outliers
age_outliers = ((df3_filtered['Age_numeric'] < 0) | (df3_filtered['Age_numeric'] > 110)).sum()
print(f"Outliers detected (age < 0 or > 110): {age_outliers}")

# Filter to reasonable age range (0-110 years)
df3_filtered['Age_cleaned'] = df3_filtered['Age_numeric'].where(
    (df3_filtered['Age_numeric'] >= 0) & (df3_filtered['Age_numeric'] <= 110),
    np.nan
)

# Create histogram
plt.figure(figsize=(10, 6))
df3_filtered['Age_cleaned'].dropna().hist(bins=30, edgecolor='black', alpha=0.7, color='coral')
plt.xlabel('Age of Victim', fontsize=12)
plt.ylabel('Frequency', fontsize=12)
plt.title('Distribution of Shark Attack Victim Ages (1940-Present)', fontsize=14, fontweight='bold')
plt.grid(True, alpha=0.3, axis='y')
plt.tight_layout()
plt.show()

# Summary statistics
print("\nAge statistics (1940-present, cleaned):")
print(df3_filtered['Age_cleaned'].describe())
print(f"Missing age values: {df3_filtered['Age_cleaned'].isna().sum()}")

# --------- GENDER ANALYSIS ---------
print("\n--- GENDER ANALYSIS ---")
# Inspect actual gender values in dataset
print(f"Unique sex values (raw, first 20): {df3_filtered['Sex'].unique()[:20]}")
print(f"Total records: {len(df3_filtered)}")
print(f"Non-null sex values: {df3_filtered['Sex'].notna().sum()}")

# Map gender values with flexible pattern matching
def map_gender(val):
    if pd.isna(val):
        return np.nan
    val_str = str(val).lower().strip()
    if any(f in val_str for f in ['f', 'female']):
        return 'Female'
    elif any(m in val_str for m in ['m', 'male']):
        return 'Male'
    else:
        return 'Unknown'

df3_filtered['Gender'] = df3_filtered['Sex'].apply(map_gender)

# Calculate proportions
gender_counts = df3_filtered['Gender'].value_counts()
print(f"\nGender distribution:")
print(gender_counts)

total_known_gender = gender_counts.sum()
male_count = gender_counts.get('Male', 0)
male_pct = (male_count / total_known_gender * 100) if total_known_gender > 0 else 0

print(f"\nProportion of victims who are male: {male_pct:.2f}%")
print(f"  ({male_count} males out of {total_known_gender} victims with known gender)")

# --------- TYPE VARIABLE CLEANING ---------
print("\n--- TYPE VARIABLE CLEANING ---")
print(f"Unique Type values (raw, before cleaning):")
print(f"  {sorted([t for t in df3_filtered['Type'].unique() if pd.notna(t)])}")

print(f"\nOriginal Type value counts:")
print(df3_filtered['Type'].value_counts(dropna=False))

# Consolidate to three categories: Provoked, Unprovoked, Unknown
def categorize_type(val):
    if pd.isna(val):
        return 'Unknown'
    val_str = str(val).lower().strip()
    if 'provoked' in val_str:
        return 'Provoked'
    elif 'unprovoked' in val_str:
        return 'Unprovoked'
    else:
        return 'Unknown'

df3_filtered['Type_cleaned'] = df3_filtered['Type'].apply(categorize_type)

print(f"\nCleaned Type value counts (consolidated to 3 categories):")
print(df3_filtered['Type_cleaned'].value_counts())

unprovoked_count = (df3_filtered['Type_cleaned'] == 'Unprovoked').sum()
unprovoked_pct = (unprovoked_count / len(df3_filtered) * 100)
print(f"\nProportion of attacks that are unprovoked: {unprovoked_pct:.2f}%")
print(f"  ({unprovoked_count} unprovoked out of {len(df3_filtered)} total attacks)")

# ============================================================================
# SUMMARY TABLE
# ============================================================================
print("\n" + "="*70)
print("DATA QUALITY SUMMARY - PHASE 1 vs PHASE 2")
print("="*70)

summary_table = pd.DataFrame({
    'Variable': ['Airbnb Price', 'Police subject_injury', 'Shark Year', 'Shark Age', 'Shark Sex', 'Shark Type'],
    'Original Missing': [
        f"{df['Price'].isna().sum()} ({(df['Price'].isna().sum()/len(df)*100):.1f}%)",
        f"{df2['subject_injury'].isna().sum()} ({(df2['subject_injury'].isna().sum()/len(df2)*100):.1f}%)",
        f"{df3['Year_numeric'].isna().sum()} ({(df3['Year_numeric'].isna().sum()/len(df3)*100):.1f}%)",
        f"{df3_filtered['Age_numeric'].isna().sum()} ({(df3_filtered['Age_numeric'].isna().sum()/len(df3_filtered)*100):.1f}%)",
        f"{df3_filtered['Sex'].isna().sum()} ({(df3_filtered['Sex'].isna().sum()/len(df3_filtered)*100):.1f}%)",
        f"{df3_filtered['Type'].isna().sum()} ({(df3_filtered['Type'].isna().sum()/len(df3_filtered)*100):.1f}%)"
    ],
    'Cleaning Action': [
        'Remove $ and commas',
        'Already clean',
        'Convert to numeric, filter 1940+',
        'Convert to numeric, filter 0-110',
        'Map to Male/Female/Unknown',
        'Consolidate to 3 categories'
    ],
    'Final Missing': [
        f"{df['Price'].isna().sum()}",
        f"{df2['subject_injury'].isna().sum()}",
        f"{df3_filtered['Year_numeric'].isna().sum()}",
        f"{df3_filtered['Age_cleaned'].isna().sum()}",
        f"{df3_filtered['Gender'].isna().sum()}",
        "0"
    ]
})

print(summary_table.to_string(index=False))
```
